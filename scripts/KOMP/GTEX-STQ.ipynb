```python
%load_ext autoreload
%autoreload 2

import os, json, sys
DIANNE = "/projects/activities/komp-histopath/USERS/domans/DIANNE"
sys.path.insert(0, f"{DIANNE}/dianne-core")
sys.path.insert(0, f"{DIANNE}/dianne-utils")
sys.path.insert(0, f"{DIANNE}/dianne-viewer")
import dianne_core
import dianne_utils as dianne
import dianne_viewer as viewer
dianne.setNotebookWidth(100)
import pandas as pd

projectDir = '/projects/activities/komp-histopath'

classifierPaths = f'{projectDir}/annotations/GTEX-round1/'
if not os.path.exists(classifierPaths):
    os.makedirs(classifierPaths)

tissues_mapping = {'kidney': ['Renal cortex',
                              'Renal medulla'],
                   'pancreas': ['Pancreas'],
                   'heart': ['Heart'],
                   'lung': ['Lung'],
                   'spleen': ['Spleen'],
                   'liver': ['Liver'],}

df_metadata_all = pd.read_csv(f'{projectDir}/USERS/domans/GTEX-normal-slides/gtex-histology-merged.csv', index_col=0)

tissue = 'kidney'

wh = df_metadata_all['PrimaryAnatomicStructures'].isin(tissues_mapping[tissue])
df_metadata = df_metadata_all.loc[wh].copy()
other_cols = [c for c in df_metadata.columns if c not in ["Pathology Categories", "Pathology Notes"]]
new_order = other_cols[:2] + ["Pathology Categories", "Pathology Notes"] + other_cols[2:]
df_metadata = df_metadata[new_order]
df_metadata = df_metadata.drop(['SpecimenDetailedDescription', 'SeriesInstanceUID', 'crdc_series_uuid', 'PathologyNotes', 'pindex'], axis=1)

dataPath = f'{projectDir}/USERS/domans/GTEX-normal-slides/results-batch-{tissue}/'
# samples = [s for s in os.listdir(dataPath) if not 'pipeline' in s][:100]
samples = ['GTEX-1A8G7-1625.oid0', 'GTEX-1B8SF-2425.oid5', 'GTEX-14JFF-2125.oid1']
sample_metadata = {s: df_metadata.loc[s.split('.oid')[0]].to_dict() for s in samples}

viewer.viewSTQkomp(dataPath, samples, load_features=True, F=2, classifierPaths=classifierPaths,
                   save_path=classifierPaths, sample_metadata=sample_metadata, mpp=0.5, username='auto');
```
